# Exercício 2 — Conta Bancária

Crie uma classe chamada `ContaBancaria`.

Cada conta deverá possuir:

- nome do titular;
- saldo.

Implemente os seguintes métodos:

- `depositar(valor)`
- `sacar(valor)`
- `consultar_saldo()`

O método `depositar()` deverá aumentar o saldo da conta.

O método `sacar()` deverá reduzir o saldo apenas quando houver saldo suficiente para realizar a operação.

O método `consultar_saldo()` deverá apresentar o saldo atual.

Crie uma conta e realize uma sequência contendo depósitos, saques e consultas de saldo.

```python
# Célula 02 - Exercício 2

class ContaBancaria:
    def __init__(self, nomeTitular: str, saldo: float) -> None:
        self.nomeTitular = nomeTitular
        self.saldo = saldo
        return

    def depositar(self) -> float:
        valor = float(input("Digite o valor que você quer depositar:\n-> "))
        self.saldo += valor
        return self.saldo

    def sacar(self) -> float:
        valor = float(input("Digite o valor que você quer sacar:\n-> "))
        self.saldo -= valor
        return self.saldo

    def consultar_saldo(self) -> None:
        return print(f"Seu saldo atual é R$ {self.saldo:.2f} reais")

conta = ContaBancaria("Henrique", 10000)

print(conta.nomeTitular, conta.depositar(), conta.sacar())
conta.consultar_saldo()
```

# Exercício 3 — Encapsulamento de uma Conta Bancária

Crie uma nova versão da classe `ContaBancaria`.

Nesta versão, o saldo não deverá ser modificado diretamente fora da classe.

As alterações de saldo deverão ocorrer somente por meio dos métodos:

- `depositar(valor)`
- `sacar(valor)`

Implemente também um método que permita consultar o saldo atual.

Crie uma conta, realize diferentes operações e demonstre o funcionamento da classe.

```python
# Célula 03 - Exercício 3

class ContaBancaria:
    def __init__(self, nome_titular: str, saldo_inicial: float) -> None:
        self.nome_titular = nome_titular
        self.__saldo = saldo_inicial 

    def depositar(self, valor: float) -> None:
        if valor > 0:
            self.__saldo += valor
            print(f"Depósito de R$ {valor:.2f} realizado com sucesso!")
        else:
            raise ValueError("O valor do depósito deve ser maior que zero.")

    def sacar(self, valor: float) -> None:
        if 0 < valor <= self.__saldo:
            self.__saldo -= valor
            print(f"Saque de R$ {valor:.2f} realizado com sucesso!")
        elif valor > self.__saldo:
            raise ValueError("Erro: Saldo insuficiente para realizar o saque.")
        else:
            raise ValueError("O valor do saque deve ser maior que zero.")

    def consultar_saldo(self) -> float:
        return self.__saldo

minha_conta = ContaBancaria("Henrique", 1000.00)

print(f"Saldo inicial: R$ {minha_conta.consultar_saldo():.2f}")

minha_conta.depositar(500.00)

minha_conta.sacar(200.00)

minha_conta.sacar(2000.00)

print(f"Saldo final: R$ {minha_conta.consultar_saldo():.2f}")
```

# Exercício 4 — Formas Geométricas

Crie uma classe abstrata chamada `Forma`.

Essa classe deverá definir o método:

`calcular_area()`

Crie então as seguintes classes:

- `Circulo`
- `Retangulo`
- `Triangulo`

Cada uma dessas classes deverá implementar o cálculo de sua própria área.

Crie pelo menos um objeto de cada tipo e apresente a área calculada.

```python
# Célula 04 - Exercício 4

from abc import ABC, abstractmethod

import math

class Forma(ABC):
    @abstractmethod
    def calcular_area() -> None:
        pass

class Circulo(Forma):
    def __init__(self, raio: float) -> None:
        self.raio = raio
        return

    def calcular_area(self) -> float:
        return math.pi*math.pow(self.raio, 2)
    
area = Circulo(5)
print(area.calcular_area())

class Retangulo(Forma):
    def __init__(self, lado: float, altura: float) -> None:
        self.lado = lado
        self.altura = altura
        return

    def calcular_area(self) -> float:
        return self.lado*self.altura

area = Retangulo(23, 20)
print(area.calcular_area())

class Triangulo(Forma):
    def __init__(self, base: float, altura: float) -> None:
        self.base = base
        self.altura = altura
        return

    def calcular_area(self) -> float:
        return (self.base*self.altura)/2

area = Triangulo(3, 6)
print(area.calcular_area())
```

# Exercício 9 — Sistema de Biblioteca

Crie uma classe chamada `Livro`.

Cada livro deverá possuir:

- título;
- autor;
- ano;
- situação de disponibilidade.

Crie também uma classe chamada `Biblioteca`.

A biblioteca deverá armazenar uma coleção de livros e disponibilizar as seguintes operações:

- adicionar um livro;
- buscar um livro pelo título;
- listar todos os livros;
- emprestar um livro;
- devolver um livro.

O sistema deverá impedir o empréstimo de um livro que já esteja emprestado.

Crie uma biblioteca com pelo menos cinco livros e demonstre todas as operações solicitadas.

```python
# Célula 09 - Exercício 9

class Livro:
    """Representa um livro."""
    def __init__(self, titulo: str, autor: str, ano: int, situacao_disponibilidade: str) -> None:
        self.titulo = titulo
        self.autor = autor
        self.ano = ano
        self.situacao_disponibilidade = situacao_disponibilidade
        return

class Biblioteca:
    """Representa uma biblioteca."""
    def __init__(self, lista_livros: list = []) -> None:
        self.lista_livros = lista_livros
        return

    def adicionar_livro(self, livro: Livro) -> list:
        self.lista_livros.append((livro.titulo, livro.autor, livro.ano, livro.situacao_disponibilidade))
        return self.lista_livros

    def listar_livros(self) -> str:
        return print(self.lista_livros)

livro = Livro("Sapiens", "Yuval Harari", 2018, "Disponivel")

biblioteca = Biblioteca()
biblioteca.adicionar_livro(livro)
biblioteca.listar_livros()
biblioteca.adicionar_livro(livro)
biblioteca.listar_livros()
```
